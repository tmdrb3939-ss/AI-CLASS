이미지 데이터는 픽셀들이 모여 하나의 형상을 이룬다. 이미지 데이터를 일반 신경망으로 처리하면 공간 정보가 사라지지만, CNN은 이미지 데이터를 보존한다.
공간정보의 보존: 이미지를 1차원으로 펼치면 상하좌우 픽셀 간의 관계를 잃게 됩니다. CNN은 2차원 구조 그대로 특징을 포착한다.
합성곱의 마법: 원본 이미지에 필터를 겹쳐 곱하는 연산으로 통해 이미지의 고유한 특징(선면, 질감 등)을 추출한다.
필터의 최적화: 필터의 값은 처음에는 무작위지만, 학습을 계속하며 데이터를 가장 잘 설명하는 최적의 형태로 진화한다.
연산 효율성:CNN은 필터를 공유하여 사용하므로, 모든 노드가 연결된 Dense레이어보다 훨씬 효율적으로 다수의 가중치를 관리한다.
stride& padding: 필터가 이동하는 보폭과 경계를 처리하는 방식으로 통해 출력 데이터의 크기와 정보 보존 여부를 세밀하게 조절한다.

정보 요약과 불변성: 풀링
합성곱을 거치면 데이터의 양이 방대해진다. 이것을 핵심정보 위주로 압축하는 과정이 필요한다.
정보의 압축: 최대 풀링은 가장 강한 특징을, 평균 풀링은 전반적인 특징을 요약하여 전달한다.
이동 불변성: 풀링을 거치면 대상의 위치가 조금 바뀌어도 동일한 특징으로 인식할 수 있는 강인함을 갖게 된다.

이미지와 달리 순서가 중요한 데이터(텍스트, 날씨 등)은 이전의 기억을 이어가는 것이 핵심이다.
RNN: 이전 단계의 정보를 다음 단계로 전달하는 루프 구조를 가진다. 문자을 읽을 때 앞 단어를 기억하며 읽는것과 같다.
LSTM: RNN의 치명적 단점인 금방 잊어버리는 문제를 해결한다. 세가지 게이트(Forget, input, output)를 통해 무엇을 버리고 무엇을 기억할지 똑똑하게 결정한다.
